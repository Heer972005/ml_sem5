```python
def hypothesis(theta_array,x):
  return (theta_array[0]+theta_array[1]*x)
```

```python
def cost_function(theta_array,x,y,m):
  total_error=0
  for i in range(m):
    total_error+=((theta_array[0]+theta_array[1]*x[i])-y[i])**2

  return (total_error/2*m)
```

```python
def gradient_descent(theta_array,x,y,m,alpha):
  sum_0,sum_1=0,0
  for i in range(m):
    sum_0+=((theta_array[0]+theta_array[1]*x[i])-y[i])
    sum_1+=x[i]*((theta_array[0]+theta_array[1]*x[i])-y[i])
  updated_theta0=theta_array[0]-(alpha/m)*sum_0
  updated_theta1=theta_array[1]-(alpha/m)*sum_1
  new_theta=[updated_theta0,updated_theta1]
  return new_theta
```

```python
def training(x,y,alpha,iters):
  theta_0,theta_1=0,0
  theta_array=[theta_0,theta_1]
  m=x.size
  cost_value=[]
  for i in range(iters):
    theta_array=gradient_descent(theta_array,x,y,m,alpha)
    cost_value.append(cost_function(theta_array,x,y,m))
    y_new=theta_array[0]+theta_array[1]*x
    plt.plot(x,y_new,'r')
    plt.scatter(x,y)
    plt.show()

  plt.figure()
  x=np.arange(0,len(cost_value),step=1)
  plt.plot(x,cost_value)
  plt.show()
  return theta_array
```

```python
clean_data=data.dropna()
clean_data.describe()
```

```python
x_train=clean_data['x']
y_train=clean_data['y']
```

```python
x_t=x_train.values.reshape(x_train.size)
y_t=y_train.values.reshape(y_train.size)
```

```python
alpha=0.0001
iter=50
# Reset the index of x_train and y_train to ensure 0-based contiguous indexing
# x_train_reset = x_train.reset_index(drop=True)
# y_train_reset = y_train.reset_index(drop=True)
theta_array=training(x_t, y_t, alpha, iter)
```
