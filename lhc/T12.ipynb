```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

points_3d = np.random.uniform(0, 1, size=(200, 3))

print(points_3d.shape)
```

```python
points_50d = np.random.uniform(0, 1, size=(200, 50))

print(points_50d.shape)
```

```python
from scipy.spatial.distance import pdist

distances_3d = pdist(points_3d, metric="euclidean")
distances_50d = pdist(points_50d, metric="euclidean")
```

```python
print(len(distances_3d))
print(len(distances_50d))
```

```python
#max distance−min distance/min distance
```

```python
ratio_3d = (
    (distances_3d.max() - distances_3d.min())
    / distances_3d.min()
)

ratio_50d = (
    (distances_50d.max() - distances_50d.min())
    / distances_50d.min()
)

print("3D ratio:", ratio_3d)
print("50D ratio:", ratio_50d)
```

```python
#Therefore, ratio became very less near to 1 in 50D space. This means that the distance between the nearest and farthest points is almost the same in high-dimensional space, which proves the curse of dimensionality.
```

```python
plt.hist(distances_3d, bins=30)
plt.xlabel("Euclidean Distance")
plt.ylabel("Frequency")
plt.title("Pairwise Euclidean Distances - 3D")
plt.show()
```

```python
#Therefore, ther is a wide spread of distances in 3D space, which means that the distance between the nearest and farthest points is significantly different.
```

```python
plt.hist(distances_50d, bins=30)
plt.xlabel("Euclidean Distance")
plt.ylabel("Frequency")
plt.title("Pairwise Euclidean Distances - 50D")
plt.show()
```

```python
#Therefore, the spread decresed which shows that the distance between the nearest and farthest points is almost the same in high-dimensional space, which proves the curse of dimensionality.
```
