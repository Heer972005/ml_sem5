```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
iris=load_iris()

X=iris.data
y=iris.target

feature_nmes=iris.feature_names
trg_nmes=iris.target_names

print("Original dataset shape:", X.shape)
print("Number of features:", X.shape[1])
print("Number of classes:", len(trg_nmes))
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)

print("\nAfter standardization:")
print("Mean of features:", X_scaled.mean(axis=0))
print("Standard deviation:", X_scaled.std(axis=0))
```

```python
pca=PCA(n_components=2)
X_pca=pca.fit_transform(X_scaled)
# transfrm --only transforms the data with hyperparamets
#fit--only th
print("\nPCA transformed shape:", X_pca.shape)

print("\nExplained variance ratio:")
print(pca.explained_variance_ratio_)

print("\nTotal variance retained:",
      pca.explained_variance_ratio_.sum())
```

```python
#Principal component plotings
plt.figure(figsize=(8, 6))

for class_id, class_name in enumerate(trg_nmes):

    plt.scatter(
        X_pca[y == class_id, 0],
        X_pca[y == class_id, 1],
        label=class_name,
        alpha=0.7
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Iris Dataset after PCA (4D → 2D)")
plt.legend()
plt.grid(True)
plt.show()
```
